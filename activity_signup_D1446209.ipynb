測試一

```python
# 固定的兩個活動名稱，以 tuple 保存。
activity_names = ("桌遊體驗", "電影欣賞")

# 收到的原始文字名單，請保留原值。
board_raw = "amy,ben,amy,cora"
movie_raw = "ben,dora,eric,ben"

# 用活動名稱查詢每人費用。
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}
```

測試一

```python
# ==================== 1. 讀取活動名稱與資料清理 ====================
# 從 activity_names 拆包讀取活動名稱
board_title, movie_title = activity_names

# 將原始文字去除兩端空白、轉為小寫，並以逗號分割成 list
# 由於原始字串中間可能含有因逗號分隔產生的多餘空白，我們先以逗號分割，再對每個姓名進行 strip() 去除空白
board_list = [name.strip().lower() for name in board_raw.split(',')]
movie_list = [name.strip().lower() for name in movie_raw.split(',')]

# ==================== 2. 使用集合去重與運算 ====================
# 將串列轉換為集合，自動去除重複元素
board_set = set(board_list)
movie_set = set(movie_list)

# 計算各種集合運算
common_set = board_set & movie_set                # 共同參加者 (交集)
all_participants_set = board_set | movie_set     # 全部不重複參加者 (聯集)
only_board_set = board_set - movie_set           # 只參加桌遊者 (差集: 桌遊 - 電影)
only_movie_set = movie_set - board_set           # 只參加電影者 (差集: 電影 - 桌遊)

# 將集合排序轉換為 list 供後續漂亮輸出
board_sorted = sorted(list(board_set))
movie_sorted = sorted(list(movie_set))
common_sorted = sorted(list(common_set))
only_board_sorted = sorted(list(only_board_set))
only_movie_sorted = sorted(list(only_movie_set))
all_participants_sorted = sorted(list(all_participants_set))

# 使用切片取得排名前兩位的預覽名單
preview_list = all_participants_sorted[:2]

# ==================== 3. 建立統計摘要字典 ====================
# 計算各項人數、人次與費用
board_count = len(board_set)
movie_count = len(movie_set)
common_count = len(common_set)
unique_count = len(all_participants_set)
register_count = board_count + movie_count

# 根據 unit_fees 取得單價並計算費用
board_fee = board_count * unit_fees[board_title]
movie_fee = movie_count * unit_fees[movie_title]
total_fee = board_fee + movie_fee

# 建立 summary 字典儲存摘要資訊
summary = {
    "board_count": board_count,
    "movie_count": movie_count,
    "common_count": common_count,
    "unique_count": unique_count,
    "register_count": register_count,
    "board_fee": board_fee,
    "movie_fee": movie_fee,
    "total_fee": total_fee
}

# ==================== 4. 輸出結果 ====================
print("HW09 活動報名摘要")
print(f"活動：{board_title}，每人 {unit_fees[board_title]:.2f} 元")
print(f"活動：{movie_title}，每人 {unit_fees[movie_title]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")
print(f"桌遊人數：{summary['board_count']} 人")
print(f"電影人數：{summary['movie_count']} 人")
print(f"共同參加者：{common_sorted}")
print(f"共同人數：{summary['common_count']} 人")
print(f"只參加桌遊者：{only_board_sorted}")
print(f"只參加電影者：{only_movie_sorted}")
print(f"全部參加者：{all_participants_sorted}")
print(f"總不重複人數：{summary['unique_count']} 人")
print(f"報名人次：{summary['register_count']} 人次")
print(f"通知名單前兩位：{preview_list}")
print(f"桌遊應收：{summary['board_fee']:.2f} 元")
print(f"電影應收：{summary['movie_fee']:.2f} 元")
print(f"總應收：{summary['total_fee']:.2f} 元")
```
